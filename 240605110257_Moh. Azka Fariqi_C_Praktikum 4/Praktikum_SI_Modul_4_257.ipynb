```python
import numpy as np

kriteria = [
    "IPK (C1)",
    "Penghasilan per tahun (C2)",
    "Jumlah tanggungan (C3)",
    "Pekerjaan orang tua (C4)"
]

alternatif = [
    "Mahadi (A1)",
    "Saimon (A2)",
    "Wahdi (A3)",
    "Aditya (A4)"
]

# Bobot setiap kriteria
wj = np.array([5, 2, 4, 2])

# Matriks keputusan
xij = np.array([
    [5, 3, 4, 4],  # A1
    [4, 4, 5, 5],  # A2
    [4, 4, 4, 4],  # A3
    [2, 4, 4, 4]   # A4
])


sum_wj = np.sum(wj)
wj_norm = np.zeros(4)
for i in range(4):
    wj_norm[i] = wj[i] / sum_wj

xij_norm = np.zeros((4, 4))

for i in range(4):
    for j in range(4):

        # Menghitung akar dari jumlah kuadrat
        pembagi = np.sqrt(
            xij[0][j] ** 2 +
            xij[1][j] ** 2 +
            xij[2][j] ** 2 +
            xij[3][j] ** 2
        )

        xij_norm[i][j] = xij[i][j] / pembagi

vij = np.zeros((4, 4))

for i in range(4):
    for j in range(4):
        vij[i][j] = xij_norm[i][j] * wj_norm[j]


conc = np.zeros((4, 4))

for i in range(4):
    for j in range(4):

        # Alternatif tidak dibandingkan dengan dirinya sendiri
        if i != j:

            for k in range(4):

                # Jika alternatif i >= alternatif j
                # maka bobot kriteria ditambahkan
                if vij[i][k] >= vij[j][k]:
                    conc[i][j] += wj_norm[k]


disc = np.zeros((4, 4))

for i in range(4):
    for j in range(4):

        if i != j:

            # Menyimpan indeks kriteria yang termasuk discordance
            disc_index = []

            for k in range(4):

                if vij[i][k] < vij[j][k]:
                    disc_index.append(k)

            # Jika terdapat kriteria discordance
            if len(disc_index) != 0:

                max_disc = []

                for index in disc_index:

                    selisih = abs(vij[i][index] - vij[j][index])
                    max_disc.append(selisih)

                # Pembilang:
                # selisih terbesar pada kriteria discordance
                pembilang = max(max_disc)

                # Penyebut:
                # selisih terbesar dari seluruh kriteria
                penyebut = max(
                    abs(vij[i][0] - vij[j][0]),
                    abs(vij[i][1] - vij[j][1]),
                    abs(vij[i][2] - vij[j][2]),
                    abs(vij[i][3] - vij[j][3])
                )

                disc[i][j] = pembilang / penyebut

            else:
                disc[i][j] = 0


c = np.sum(conc) / (4 * (4 - 1))
d = np.sum(disc) / (4 * (4 - 1))

fij = np.zeros((4, 4))

for i in range(4):
    for j in range(4):

        if conc[i][j] >= c:
            fij[i][j] = 1
        else:
            fij[i][j] = 0


gij = np.zeros((4, 4))

for i in range(4):
    for j in range(4):

        if disc[i][j] <= d:
            gij[i][j] = 1
        else:
            gij[i][j] = 0


print("=" * 60)
print("SISTEM PENDUKUNG KEPUTUSAN")
print("SELEKSI PENERIMA BEASISWA METODE ELECTRE")
print("=" * 60)


print("\nDaftar Kriteria")
for i in range(4):
    print(f"Kriteria {i + 1}: {kriteria[i]}")


print("\nBobot Kriteria")
for i in range(4):
    print(f"Kriteria {i + 1}: {wj[i]}")


print("\nDaftar Alternatif")
for i in range(4):
    print(f"Alternatif {i + 1}: {alternatif[i]}")


print("\nMatriks Keputusan")
print(xij)

print("\nBobot Ternormalisasi")
print(wj_norm)

print("\nMatriks Keputusan Ternormalisasi")
print(xij_norm)

print("\nMatriks Keputusan Ternormalisasi Terbobot")
print(vij)

print("\nConcordance Matrix")
print(conc)

print("\nDiscordance Matrix")
print(disc)

print("\nThreshold Concordance (c)")
print(c)

print("\nThreshold Discordance (d)")
print(d)

print("\nMatriks Dominan Concordance (F)")
print(fij)

print("\nMatriks Dominan Discordance (G)")
print(gij)
```

```python

eij = fij * gij

print("\nAggregate Dominance matrix")
print(eij)

skor_akhir = np.sum(eij, axis=1)

print("\nSkor Akhir")

for i in range(4):
    print(f"{alternatif[i]}: {int(skor_akhir[i])}")
```
