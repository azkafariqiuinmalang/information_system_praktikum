```python
import numpy as np
import pandas as pd

# Data alternatif
alternatif = ["A1", "A2", "A3", "A4", "A5"]

# Matriks keputusan
X = np.array([
    [1, 1, 3, 4, 1, 4],
    [1, 1, 4, 3, 1, 3],
    [1, 1, 4, 4, 1, 4],
    [1, 1, 3, 3, 0, 3],
    [1, 1, 3, 3, 1, 4]
], dtype=float)

# Bobot kriteria
bobot = np.array([0.290, 0.173, 0.091, 0.162, 0.080, 0.204])

normalisasi = np.zeros_like(X)

for i in range(len(X)):
    pembagi = np.sqrt(np.sum(X[i] ** 2))
    normalisasi[i] = X[i] / pembagi

terbobot = normalisasi * bobot




benefit = [0, 1, 2, 4, 5]

cost = [3]

nilai_benefit = np.sum(terbobot[:, benefit], axis=1)
nilai_cost = np.sum(terbobot[:, cost], axis=1)

Yi = nilai_benefit - nilai_cost


hasil = pd.DataFrame({
    "Alternatif": alternatif,
    "Max (Benefit)": nilai_benefit,
    "Min (Cost)": nilai_cost,
    "Yi": Yi
})

hasil["Ranking"] = hasil["Yi"].rank(
    ascending=False,
    method="min"
).astype(int)

hasil = hasil.sort_values("Ranking")

print("HASIL MOORA")
print("Benefit : C1,C2,C3,C5")
print("Cost    : C4")

print("\nHASIL PERANGKINGAN")
print(hasil.round(6).to_string(index=False))
```
