```python
import math


# Fungsi keanggotaan menurun: TURUN atau SEDIKIT
def keanggotaan_turun(nilai, batas_min, batas_max):
    if nilai <= batas_min:
        return 1.0
    if nilai >= batas_max:
        return 0.0
    return (batas_max - nilai) / (batas_max - batas_min)


# Fungsi keanggotaan meningkat: NAIK atau BANYAK
def keanggotaan_naik(nilai, batas_min, batas_max):
    if nilai <= batas_min:
        return 0.0
    if nilai >= batas_max:
        return 1.0
    return (nilai - batas_min) / (batas_max - batas_min)


# 1. Data kasus
permintaan = 1800
persediaan = 250

permintaan_min = 500
permintaan_max = 2000

persediaan_min = 100
persediaan_max = 300


# 2. Fuzzifikasi
pmt_turun = keanggotaan_turun(
    permintaan, permintaan_min, permintaan_max
)
pmt_naik = keanggotaan_naik(
    permintaan, permintaan_min, permintaan_max
)

psd_sedikit = keanggotaan_turun(
    persediaan, persediaan_min, persediaan_max
)
psd_banyak = keanggotaan_naik(
    persediaan, persediaan_min, persediaan_max
)


# 3. Inferensi: alpha-predikat menggunakan MIN untuk operator AND

# R1: Permintaan TURUN dan Persediaan BANYAK
alpha1 = min(pmt_turun, psd_banyak)
z1 = permintaan - persediaan

# R2: Permintaan TURUN dan Persediaan SEDIKIT
alpha2 = min(pmt_turun, psd_sedikit)
z2 = permintaan

# R3: Permintaan NAIK dan Persediaan BANYAK
alpha3 = min(pmt_naik, psd_banyak)
z3 = permintaan

# R4: Permintaan NAIK dan Persediaan SEDIKIT
alpha4 = min(pmt_naik, psd_sedikit)
z4 = 1.25 * permintaan - persediaan


# 4. Defuzzifikasi dengan rata-rata terbobot
pembilang = (
    alpha1 * z1
    + alpha2 * z2
    + alpha3 * z3
    + alpha4 * z4
)

penyebut = alpha1 + alpha2 + alpha3 + alpha4

if penyebut == 0:
    raise ValueError("Tidak ada aturan fuzzy yang aktif.")

hasil = pembilang / penyebut

# Pembulatan ke atas sesuai contoh hasil pada modul
produksi = math.ceil(hasil)


# 5. Menampilkan hasil
print("=" * 55)
print("       PERHITUNGAN METODE FUZZY SUGENO")
print("=" * 55)

print("\nDATA KASUS")
print(f"Permintaan          : {permintaan} buah/hari")
print(f"Persediaan          : {persediaan} buah")

print("\nHASIL FUZZIFIKASI")
print(f"Permintaan TURUN    : {pmt_turun:.6f}")
print(f"Permintaan NAIK     : {pmt_naik:.6f}")
print(f"Persediaan SEDIKIT  : {psd_sedikit:.6f}")
print(f"Persediaan BANYAK   : {psd_banyak:.6f}")

print("\nHASIL INFERENSI")
print(f"R1: alpha = {alpha1:.6f}, z1 = {z1:.2f}")
print(f"R2: alpha = {alpha2:.6f}, z2 = {z2:.2f}")
print(f"R3: alpha = {alpha3:.6f}, z3 = {z3:.2f}")
print(f"R4: alpha = {alpha4:.6f}, z4 = {z4:.2f}")

print("\nHASIL DEFUZZIFIKASI")
print(f"Jumlah alpha * z    : {pembilang:.6f}")
print(f"Jumlah alpha        : {penyebut:.6f}")
print(f"Hasil produksi      : {hasil:.4f} buah/hari")
print(f"Dibulatkan ke atas  : {produksi} buah/hari")

print("\nKESIMPULAN")
print(
    f"Dengan permintaan {permintaan} buah/hari dan "
    f"persediaan {persediaan} buah, jumlah produksi "
    f"yang disarankan adalah {produksi} buah/hari."
)
```
