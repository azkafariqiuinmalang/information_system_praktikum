```python
import numpy as np
import pandas as pd

# Data alternatif
alternatif = ["Taufan", "Bunga", "Yoga", "Desi"]

# Matriks keputusan
X = np.array([
    [5, 5, 3],
    [5, 2, 3],
    [2, 5, 5],
    [5, 2, 5]
], dtype=float)

# Bobot kriteria
bobot = np.array([50, 20, 30], dtype=float)

# Menampilkan matriks keputusan
df_x = pd.DataFrame(
    X,
    index=alternatif,
    columns=["C1", "C2", "C3"]
)

print("=== Matriks Keputusan ===")
print(df_x)

# 1. Normalisasi matriks keputusan
normalisasi = X / np.sqrt(np.sum(X**2, axis=0))

df_normalisasi = pd.DataFrame(
    normalisasi,
    index=alternatif,
    columns=["C1", "C2", "C3"]
)

print("\n=== Matriks Keputusan Ternormalisasi ===")
print(df_normalisasi.round(6))

# 2. Matriks normalisasi terbobot
terbobot = normalisasi * bobot

df_terbobot = pd.DataFrame(
    terbobot,
    index=alternatif,
    columns=["C1", "C2", "C3"]
)

print("\n=== Matriks Normalisasi Terbobot ===")
print(df_terbobot.round(6))

# 3. Solusi ideal positif dan negatif
# Semua kriteria merupakan Benefit
ideal_positif = np.max(terbobot, axis=0)
ideal_negatif = np. min(terbobot, axis=0)

print("\n=== Solusi Ideal ===")
print("Ideal Positif :", np.round(ideal_positif, 6))
print("Ideal Negatif :", np.round(ideal_negatif, 6))

# 4. Menghitung jarak terhadap solusi ideal
D_plus = np.sqrt(np.sum((terbobot - ideal_positif)**2, axis=1))
D_minus = np.sqrt(np.sum((terbobot - ideal_negatif)**2, axis=1))

# 5. Menghitung nilai preferensi
V = D_minus / (D_plus + D_minus)

# Menampilkan hasil
hasil = pd.DataFrame({
    "Alternatif": alternatif,
    "D+": D_plus,
    "D-": D_minus,
    "Nilai Preferensi": V
})

hasil["Ranking"] = hasil["Nilai Preferensi"].rank(
    ascending=False,
    method="min"
).astype(int)

hasil = hasil.sort_values("Ranking")

print("\n=== Hasil Akhir TOPSIS ===")
print(hasil.round(6).to_string(index=False))
```

```python
import numpy as np
import pandas as pd

# Data alternatif
alternatif = ["Taufan", "Bunga", "Yoga", "Desi"]

# Matriks keputusan
X = np.array([
    [5, 5, 3],
    [5, 2, 3],
    [2, 5, 5],
    [5, 2, 5]
], dtype=float)

# Bobot kriteria
bobot = np.array([50, 20, 30], dtype=float)

# Menampilkan matriks keputusan
df_x = pd.DataFrame(
    X,
    index=alternatif,
    columns=["C1", "C2", "C3"]
)

print("=== Matriks Keputusan ===")
print(df_x)

# 1. Normalisasi matriks keputusan
normalisasi = X / np.sqrt(np.sum(X**2, axis=0))

# 2. Matriks normalisasi terbobot
terbobot = normalisasi * bobot

print("\n=== Matriks Normalisasi Terbobot ===")
print(
    pd.DataFrame(
        terbobot,
        index=alternatif,
        columns=["C1", "C2", "C3"]
    ).round(6)
)

# C1 = Cost
# C2 = Cost
# C3 = Benefit

ideal_positif = np.array([
    np.min(terbobot[:, 0]),  # C1 Cost
    np.min(terbobot[:, 1]),  # C2 Cost
    np.max(terbobot[:, 2])   # C3 Benefit
])

ideal_negatif = np.array([
    np.max(terbobot[:, 0]),  # C1 Cost
    np.max(terbobot[:, 1]),  # C2 Cost
    np.min(terbobot[:, 2])   # C3 Benefit
])

print("\n=== Solusi Ideal ===")
print("Ideal Positif :", np.round(ideal_positif, 6))
print("Ideal Negatif :", np.round(ideal_negatif, 6))

# 3. Menghitung jarak terhadap solusi ideal
D_plus = np.sqrt(np.sum((terbobot - ideal_positif)**2, axis=1))
D_minus = np.sqrt(np.sum((terbobot - ideal_negatif)**2, axis=1))

# 4. Menghitung nilai preferensi
V = D_minus / (D_plus + D_minus)

# 5. Menampilkan hasil
hasil = pd.DataFrame({
    "Alternatif": alternatif,
    "D+": D_plus,
    "D-": D_minus,
    "Nilai Preferensi": V
})

hasil["Ranking"] = hasil["Nilai Preferensi"].rank(
    ascending=False,
    method="min"
).astype(int)

hasil = hasil.sort_values("Ranking")

print("\n=== Hasil TOPSIS C1 dan C2 sebagai Cost ===")
print(hasil.round(6).to_string(index=False))
```
