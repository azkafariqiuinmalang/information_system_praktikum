```python

import numpy as np

def turun_or_sedikit(b, a, x):
    if x <= a:
        nilai = 1
    elif a < x < b:
        nilai = (b - x) / (b - a)
    else:
        nilai = 0

    return nilai


def naik_or_banyak(b, a, x):
    if x <= a:
        nilai = 0
    elif a < x < b:
        nilai = (x - a) / (b - a)
    else:
        nilai = 1

    return nilai

def agregasi_turun(b, a, alfa):
    nilai = b - (alfa * (b - a))
    return nilai


def agregasi_naik(b, a, alfa):
    nilai = alfa * (b - a) + a
    return nilai


permintaan_tertinggi = 2000
permintaan_terendah = 500

persediaan_tertinggi = 300
persediaan_terendah = 100

produksi_tertinggi = 4000
produksi_terendah = 500

permintaan_saat_ini = 1800
persediaan_saat_ini = 250

alfa = []
z = []


print("========== PERHITUNGAN PRODUKSI BARANG DENGAN METODE FUZZY TSUKAMOTO ==========")
print()

print("Suatu perusahaan akan memproduksi powerbank jenis XZ.")
print("Dari data 2 bulan terakhir, permintaan terbesar mencapai 2000 buah/hari")
print("dan yang terkecil 500 buah/hari.")
print("Persediaan di gudang, terbanyak mencapai 300 buah/hari")
print("dan terkecil 100 buah/hari.")
print("Perusahaan mampu memproduksi maksimal 4000 buah/hari,")
print("dan untuk efisiensi mesin minimal harus memproduksi 500 buah/hari.")
print()
print("Permintaan saat ini: 1800 buah")
print("Persediaan saat ini: 250 buah")
print()

print("Aturan fuzzy yang digunakan:")
print("R1: IF Permintaan TURUN AND Persediaan BANYAK THEN Produksi BERKURANG")
print("R2: IF Permintaan TURUN AND Persediaan SEDIKIT THEN Produksi BERKURANG")
print("R3: IF Permintaan NAIK AND Persediaan BANYAK THEN Produksi BERTAMBAH")
print("R4: IF Permintaan NAIK AND Persediaan SEDIKIT THEN Produksi BERTAMBAH")

print()
print("====== AGREGASI BERDASARKAN ATURAN FUZZY ======")
print()


# ============================================================
# ATURAN R1
# IF Permintaan TURUN AND Persediaan BANYAK
# THEN Produksi BERKURANG
# ============================================================

print("R1: IF permintaan turun AND persediaan banyak THEN produksi berkurang.")

alfa_R1 = min(
    turun_or_sedikit(
        permintaan_tertinggi,
        permintaan_terendah,
        permintaan_saat_ini
    ),
    naik_or_banyak(
        persediaan_tertinggi,
        persediaan_terendah,
        persediaan_saat_ini
    )
)

alfa.append(alfa_R1)

z_R1 = agregasi_turun(
    produksi_tertinggi,
    produksi_terendah,
    alfa_R1
)

z.append(z_R1)

print("Alfa :", alfa_R1)
print("z    :", z_R1)
print()


# ============================================================
# ATURAN R2
# IF Permintaan TURUN AND Persediaan SEDIKIT
# THEN Produksi BERKURANG
# ============================================================

print("R2: IF permintaan turun AND persediaan sedikit THEN produksi berkurang.")

alfa_R2 = min(
    turun_or_sedikit(
        permintaan_tertinggi,
        permintaan_terendah,
        permintaan_saat_ini
    ),
    turun_or_sedikit(
        persediaan_tertinggi,
        persediaan_terendah,
        persediaan_saat_ini
    )
)

alfa.append(alfa_R2)

z_R2 = agregasi_turun(
    produksi_tertinggi,
    produksi_terendah,
    alfa_R2
)

z.append(z_R2)

print("Alfa :", alfa_R2)
print("z    :", z_R2)
print()


# ============================================================
# ATURAN R3
# IF Permintaan NAIK AND Persediaan BANYAK
# THEN Produksi BERTAMBAH
# ============================================================

print("R3: IF permintaan naik AND persediaan banyak THEN produksi bertambah.")

alfa_R3 = min(
    naik_or_banyak(
        permintaan_tertinggi,
        permintaan_terendah,
        permintaan_saat_ini
    ),
    naik_or_banyak(
        persediaan_tertinggi,
        persediaan_terendah,
        persediaan_saat_ini
    )
)

alfa.append(alfa_R3)

z_R3 = agregasi_naik(
    produksi_tertinggi,
    produksi_terendah,
    alfa_R3
)

z.append(z_R3)

print("Alfa :", alfa_R3)
print("z    :", z_R3)
print()


# ============================================================
# ATURAN R4
# IF Permintaan NAIK AND Persediaan SEDIKIT
# THEN Produksi BERTAMBAH
# ============================================================

print("R4: IF permintaan naik AND persediaan sedikit THEN produksi bertambah.")

alfa_R4 = min(
    naik_or_banyak(
        permintaan_tertinggi,
        permintaan_terendah,
        permintaan_saat_ini
    ),
    turun_or_sedikit(
        persediaan_tertinggi,
        persediaan_terendah,
        persediaan_saat_ini
    )
)

alfa.append(alfa_R4)

z_R4 = agregasi_naik(
    produksi_tertinggi,
    produksi_terendah,
    alfa_R4
)

z.append(z_R4)

print("Alfa :", alfa_R4)
print("z    :", z_R4)
print()


# ============================================================
# DEFUZZIFIKASI
# ============================================================

df = 0

for i in range(len(alfa)):
    df += alfa[i] * z[i]

defuz = np.ceil(df / sum(alfa))


# ============================================================
# HASIL AKHIR
# ============================================================

print("====== DEFUZZIFIKASI ======")
print()
print("Jadi, nilai produksi barang adalah", defuz)
```
